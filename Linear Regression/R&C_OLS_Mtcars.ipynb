```python
import statsmodels.api as sm
```

```python
# Fetch the dataset from R's core datasets repository
mtcars = sm.datasets.get_rdataset("mtcars", "datasets").data

print(mtcars.head())
```

```python
import numpy as np
```

```python
import pandas as pd
```

```python
df_subset = mtcars[['hp', 'mpg']]
```

```python
df_subset.head()
```

```python
df_final = df_subset.reset_index(drop=True)
df_final.head()
```

```python
y = df_final['mpg'].to_numpy()
X_raw = df_final['hp'].to_numpy()
```

```python
X = np.vander(X_raw, 2) # creates a column of 1's in X_raw
```

```python
beta = np.linalg.inv(X.T @ X) @ X.T @ y
```

```python
slope = beta[0]
intercept = beta[1]

print(f"Intercept: {intercept:.4f}")
print(f"Slope (hp): {slope:.4f}")
print(f"\nPrediction Equation: mpg = {intercept:.4f} + ({slope:.4f} * hp)")
```

```python
# 1. Prepare your data
X = df_final['hp']
y = df_final['mpg']

# 2. Statsmodels requires explicitly adding a constant (intercept) column
X = sm.add_constant(X)
# 3. Fit the OLS model
model = sm.OLS(y, X).fit()

# 4. Print the full regression analysis report
print(model.summary())
```
